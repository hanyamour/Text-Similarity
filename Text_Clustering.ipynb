```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from IPython.display import display, HTML
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from scipy.spatial.distance import cdist

data = pd.read_csv("quanta_physics_preprocessing.csv")

#Menghapus data dengan teks kosong
data = data.dropna(subset=["Processed_text"]).reset_index(drop=True)
data["Processed_text"] = data["Processed_text"].astype(str)
data = data[data["Processed_text"].str.strip() != ""].reset_index(drop=True)

#Menggunakan hasil preprocessing sebelumnya
corpus = data["Processed_text"].tolist()
print("Total rows:", len(data))
display(HTML(data.head().to_html()))
```

Jumlah data yang digunakan adalah 775 artikel.


Cosine Similarity menggunakan TF-IDF

```python
#Membuat TF-IDF Vectorizer
vectorizer = TfidfVectorizer(
    sublinear_tf=True,
    max_features=5000
)

#Mengubah corpus menjadi matriks TF-IDF
documents_tfidf_features = vectorizer.fit_transform(corpus)

#Menampilkan jumlah fitur
print("Total features:",
      vectorizer.get_feature_names_out().shape)
vectorizer.get_feature_names_out()[:100]
```

TF-IDF digunakan untuk mengubah teks menjadi representasi numerik berdasarkan tingkat kepentingan setiap kata dalam dokumen.

Kata yang lebih khas pada suatu artikel dapat memiliki bobot lebih tinggi dibandingkan kata yang sering muncul pada banyak artikel.

```python
from sklearn.metrics.pairwise import cosine_similarity
TopK = 10

#Query pencarian
query = "quantum physics space time gravity black hole"

#Mengubah query menjadi TF-IDF vector
query_tfidf_features = vectorizer.transform([query])
print("TF-IDF features for query",
      query_tfidf_features, "\n")

#Menghitung Cosine Similarity
similarities = cosine_similarity(
    documents_tfidf_features,
    query_tfidf_features
)

#Mengambil 10 dokumen dengan similarity tertinggi
top_indices = similarities[:, 0].argsort()[::-1][:TopK]

#Menampilkan dokumen yang relevan
titles = []
abstracts = []
similarity_scores = []

for doc_index in top_indices:

    titles.append(data.iloc[doc_index].Title)

    abstracts.append(
        data.iloc[doc_index].Summary.replace("\n", " ")
    )

    similarity_scores.append(
        similarities[doc_index, 0]
    )

#Membuat tabel hasil
results_cosine = pd.DataFrame({
    "Title": titles,
    "Abstract": abstracts,
    "Similarities": similarity_scores
})

display(HTML(results_cosine.to_html()))
```

Cosine Similarity digunakan untuk menghitung tingkat kemiripan antara query dengan setiap dokumen berdasarkan sudut antara dua vektor.

Semakin mendekati nilai 1, semakin mirip isi dokumen dengan query. Dokumen kemudian diurutkan berdasarkan nilai similarity tertinggi dan diambil 10 artikel teratas.

Jaccard Similarity

```python
from sklearn.feature_extraction.text import CountVectorizer

#Vectorize corpus menggunakan Binary CountVectorizer
vectorizer = CountVectorizer(
    binary=True,
    max_features=5000
)

documents_binarized_features = vectorizer.fit_transform(corpus).toarray()

#Menampilkan jumlah fitur
print("Total features:",
      vectorizer.get_feature_names_out().shape)

vectorizer.get_feature_names_out()[:100]
```

Binary CountVectorizer mengubah teks menjadi representasi biner, yaitu 1 dan 0.

Nilai 1 menunjukkan bahwa kata terdapat dalam dokumen, sedangkan nilai 0 menunjukkan bahwa kata tidak terdapat dalam dokumen. Kemunculan kata berulang tetap dihitung sebagai 1.

```python
from scipy.spatial.distance import cdist

#Mengubah query menjadi binary vector
TopK = 10
query = "quantum physics space time gravity black hole"
query_binarized_features = vectorizer.transform([query]).toarray()

#Menghitung Jaccard Similarity
similarities = 1 - cdist(
    documents_binarized_features,
    query_binarized_features,
    metric="jaccard"
)

#Mengambil Top 10 dokumen dengan similarity tertinggi
top_indices = similarities[:, 0].argsort()[::-1][:TopK]

#Menampilkan dokumen yang relevan
titles = []
abstracts = []
similarity_scores = []

for doc_index in top_indices:

    titles.append(data.iloc[doc_index].Title)

    abstracts.append(
        data.iloc[doc_index].Summary.replace("\n", " ")
    )

    similarity_scores.append(
        similarities[doc_index, 0]
    )

#Membuat DataFrame hasil
results_jaccard = pd.DataFrame({
    "Title": titles,
    "Abstract": abstracts,
    "Similarities": similarity_scores
})

display(HTML(results_jaccard.to_html()))
```

Jaccard Similarity menghitung kemiripan berdasarkan perbandingan jumlah kata yang sama antara query dan dokumen dengan jumlah keseluruhan kata unik dari keduanya.

Membandingkan hasil Cosine dan Jaccard

```python
#Membandingkan artikel yang muncul pada kedua metode

cosine_titles = set(results_cosine["Title"])
jaccard_titles = set(results_jaccard["Title"])

common_titles = cosine_titles.intersection(jaccard_titles)

print("Jumlah artikel Top 10 Cosine:",
      len(cosine_titles))

print("Jumlah artikel Top 10 Jaccard:",
      len(jaccard_titles))

print("Jumlah artikel yang sama:",
      len(common_titles))

print("\nArtikel yang muncul pada kedua metode:")

for title in results_cosine["Title"]:
    if title in common_titles:
        print("-", title)
```

**Analisis Cosine Similarity:**

Berdasarkan hasil perhitungan, artikel The Physicist Who’s Challenging the Quantum Orthodoxy memperoleh nilai similarity tertinggi sebesar 0.4283. Hal ini menunjukkan bahwa artikel tersebut memiliki kemiripan paling tinggi dengan query yang digunakan. Artikel lainnya juga membahas topik yang berkaitan dengan quantum physics, black holes, dan space-time.

**Analisis Jaccard Similarity:**

Berdasarkan hasil perhitungan, artikel The Physicist Who’s Challenging the Quantum Orthodoxy juga memperoleh nilai similarity tertinggi sebesar 0.3077. Hal ini menunjukkan bahwa artikel tersebut memiliki banyak kesamaan kata dengan query yang digunakan.

**Perbandingan dan Kesimpulan:**

Kedua metode menghasilkan artikel dengan peringkat pertama yang sama, tetapi urutan artikel lainnya berbeda. Cosine Similarity memperhitungkan bobot kata menggunakan TF-IDF, sedangkan Jaccard Similarity hanya memperhitungkan keberadaan kata yang sama.

Dari hasil tersebut, kedua metode dapat digunakan untuk menemukan artikel yang relevan berdasarkan query, tetapi menghasilkan nilai dan urutan kemiripan yang berbeda.
